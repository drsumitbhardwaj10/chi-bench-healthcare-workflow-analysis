# CHI-Bench Prior Authorization E2E Workflow Analysis

This notebook analyzes the structure of the 23 provider-to-payer prior authorization E2E workflows in CHI-Bench. The analysis uses task instructions and task metadata only and does not inspect hidden evaluation artifacts.

## Objective

Analyze E2E workflow coverage, condition diversity, execution requirements, and the presence of both provider- and payer-side interfaces.

```python
from pathlib import Path
import csv
from collections import Counter

DATA = Path('../data/pa_e2e_task_metadata.csv')

with DATA.open(encoding='utf-8', newline='') as f:
    rows = list(csv.DictReader(f))

print('Workflows:', len(rows))
print('Distinct conditions:', len(set(r['condition'] for r in rows)))
```

## Condition diversity

The E2E benchmark contains multiple clinical scenarios rather than concentrating on a single condition.

```python
condition_counts = Counter(r['condition'] for r in rows)

for condition, count in condition_counts.most_common():
    print(f'{condition}: {count}')
```

## Execution requirements

Every E2E workflow exposes both provider and payer MCP interfaces. The longer agent timeout reflects the multi-party workflow structure.

```python
print('Provider MCP availability:')
print(Counter(r['provider_mcp'] for r in rows))

print('\nPayer MCP availability:')
print(Counter(r['payer_mcp'] for r in rows))

print('\nVerifier timeout:')
print(Counter(r['verifier_timeout_sec'] for r in rows))

print('\nAgent timeout:')
print(Counter(r['agent_timeout_sec'] for r in rows))
```

## Key observations

The E2E benchmark contains 23 provider-to-payer workflows spanning 20 distinct condition titles. Every workflow exposes both provider and payer MCP interfaces, making the benchmark structurally different from single-actor provider or payer workflows. The 2,700-second agent timeout is three times the 900-second timeout used by the provider and payer workflow families, consistent with the greater execution scope of an end-to-end interaction.
